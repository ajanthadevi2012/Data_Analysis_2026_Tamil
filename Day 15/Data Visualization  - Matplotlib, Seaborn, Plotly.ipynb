## 3. Bar chart

```python
data_1 = {"C":20,'C++':15, "Java":30,"Python":55}
courses = list(data_1.keys())
values = list(data_1.values())
fig = plt.figure(figsize = (8,5))

plt.bar(courses,values, color = 'maroon', width = 0.4) # to draw the chart

plt.xlabel("courses offered")
plt.ylabel("no. of students enrolled")
plt.title("students enrolled in different courses")

plt.show()

```

```python
# height of the bar
IT = [22,23,34,45,66]
ECE = [55,45,43,32,21]
CSE = [44,55,66,33,22]

# width
barwidth = 0.25
fig = plt.subplots(figsize = (8,5))

# position
br1 = np.arange(len(IT))
br2 = [x+barwidth for x in br1]
br3 = [x+barwidth for x in br2]

# plot

plt.bar(br1, IT, color = 'r',width =barwidth, edgecolor = 'grey', label = 'IT' )
plt.bar(br2, ECE, color = 'g',width =barwidth, edgecolor = 'grey', label = 'ECE' )
plt.bar(br3, CSE, color = 'b',width =barwidth, edgecolor = 'grey', label = 'CSE' )

# add xticks
plt.xlabel("Branch", fontweight = 'bold', fontsize = 15)
plt.ylabel("Students, Passed",fontweight = 'bold', fontsize = 15 )
plt.xticks([r+barwidth for r in range(len(IT))], ['2020','2021','2022','2023','2024'])

plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

N = 5
boys = (40,35,30,35,27)
girls = (25,32,34,20,25)
boy_std = (2,3,4,1,2)
girls_std = (3,5,2,3,3)

ind = np.arange(N)
width = 0.35
fig = plt.subplots(figsize = (8,5))
p1 = plt.bar(ind,boys, width, yerr = boy_std)
p2 = plt.bar(ind,girls, width, yerr = girls_std)

plt.ylabel("Contribution")
plt.xlabel("Contribution by the Team")

plt.xticks(ind,('T1','T2','T3','T4','T5'))
plt.yticks(np.arange(0,81,10))
plt.legend((p1[0],p2[0]),("Boys",'Girls'))
plt.show()
```

```python
# creating dropdown menu
import plotly.graph_objects as px
plot = px.Figure(data = [px.Scatter(x = data['day'], y = data['tip'], mode = 'markers')])

# Add dropdown
plot.update_layout(updatemenus=[dict(buttons=list([dict(args=["type", "scatter"],
                    label="Scatter Plot",method="restyle"),
                dict(args=["type", "bar"],label="Bar Chart",method="restyle")]),direction="down",),])
plot.show()
```

```python
# Add dropdown
plot.update_layout(
    updatemenus=[dict(type="buttons",direction="left",buttons=list([dict(args=["type", "scatter"],
                    label="Scatter Plot",method="restyle"), dict(args=["type", "bar"],
                    label="Bar Chart",method="restyle")]),),])
plot.show()
```
