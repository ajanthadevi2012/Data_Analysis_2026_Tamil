# 5 Bar Graph  <a class="anchor" id="5"></a>

[Table of Contents](#0.1)


```python
## Basic Plot
plt.figure(figsize=(8,8))
id1 = np.arange(1,10)
salary = np.arange(20,110,10)
plt.bar(id1,salary)
plt.xlabel('Employee ID')
plt.ylabel('Salary')
plt.show()
```

```python
id1 = np.arange(1,10)
salary = np.arange(20,110,10)
plt.figure(figsize=(8,5)) # Setting the figure size
ax = plt.axes()
ax.set_facecolor("#ECF0F1") # Setting the background color by specifying the HEX Code
plt.bar(id1,salary,color = '#FFA726')
plt.xlabel(r'$Employee $ $ ID$')
plt.ylabel(r'$Salary$')
plt.show()
```

```python
#Plotting multiple sets of data
x1= [1,3,5,7]
x2=[2,4,6,8]
y1 = [7,7,7,7]
y2= [17,18,29,40]
plt.figure(figsize=(8,6))
ax = plt.axes()
ax.set_facecolor("white")
plt.bar(x1,y1,label = "First",color = '#42B300') # First set of data
plt.bar(x2,y2,label = "Second",color = '#94E413') # Second set of data
plt.xlabel('$X$')
plt.ylabel('$Y$')
plt.title ('$Bar $ $ Chart$')
plt.legend()
plt.show()
```

## 5.1 How to show values and label at the top of the graph <a class="anchor" id="5.1"></a>

[Table of Contents](#0.1)


```python
barplot = plt.bar(id1, salary)
for bar in barplot:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval, int(yval), va='bottom') #va: vertical alignment y positional argument
    
plt.title(" Bar graph example")
plt.xlabel('Employees')
plt.ylabel("Salary")
```

## 5.2 How to increase the size of the graph  <a class="anchor" id="5.2"></a>

```python
plt.figure(figsize=(7,7))
barplot = plt.bar(id1, salary)
for bar in barplot:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval, int(yval), va='bottom') #va: vertical alignment y positional argument
    
plt.title(" Bar graph example")
plt.xlabel('Employees')
plt.ylabel("Salary")
plt.show()
```

## 5.3 How to hide axis  <a class="anchor" id="5.3"></a>

[Table of Contents](#0.1)

Sometimes we hide y-axis to give aesthetic touch to our bar graph. To do this in matplotlib, we can leverage plt.yticks( ) function.[ ] means empty list.[](http://)

```python
plt.figure(figsize=(7,7))
barplot = plt.bar(id1, salary)
for bar in barplot:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2.0, yval, int(yval), va='bottom') #va: vertical alignment y positional argument
    plt.yticks([])
    #plt.xticks([])
plt.title(" Bar graph example")
plt.xlabel('Employees')
plt.ylabel("Salary")
plt.show()
```

## 5.5 How to sort or order bars <a class="anchor" id="5.5"></a>

[Table of Contents](#0.1)

We can arrange the bars based on the values. We need to combine both the lists and then sort them based on value of list y. zip( ) function is used to combine items of lists x and y. sorted( )sort them based on y. Then we split and store it in x and y lists.

```python
plt.figure(figsize=(8,8))
y,x = zip(*sorted(zip(salary,id1)))
plt.barh(x,y)
```

### Reverse order of bars

plt.subplot() is used to find out current axes and then invert function assists to reverse the order.

```python
plt.barh(x,y)
ax=plt.subplot()
ax.invert_yaxis()
```

## 5.8 Grouped Bar Charts <a class="anchor" id="5.8"></a>

[Table of Contents](#0.1)

```python
# Grouped Bar Chart
plt.figure(figsize=(7,9))
# set width of bar
barWidth = 0.25
# set height of bar
y1= np.array([17,18,29,40])
y2 =np.array([20,21,22,23])
y3 =np.array([5,9,11,12])
# Set position of bar on X axis
pos1 = np.arange(len(y1))
pos2 = [x + barWidth for x in pos1]
pos3 = [x + barWidth for x in pos2]
# Make the plot
plt.bar(pos1, y1, color='#FBC02D', width=barWidth, label='Batting')
plt.bar(pos2, y2, color='#F57F17', width=barWidth, label='Bowling')
plt.bar(pos3, y3, color='#E65100', width=barWidth, label='Fielding')
# Add xticks on the middle of the group bars
plt.xlabel('Assignee', fontweight='bold')
plt.ylabel('Number of Tickets', fontweight='bold')
plt.xticks([i + barWidth for i in range(len(y1))], ['Sachin', 'Saurav', 'Rahul', 'Robin'])

# Create legend & Show graphic
plt.legend()
plt.show()
np.arange(len(y1))
```

##  Contour Plots 


```python
def f(x, y):
    return (1 - x / 2 + x ** 5 + y ** 3) * np.exp(-x ** 2 -y ** 2)
n = 256
x = np.linspace(-3, 3, n)
y = np.linspace(-3, 3, n)
X, Y = np.meshgrid(x, y)
plt.contourf(X, Y, f(X, Y), 8, alpha=.75, cmap='jet')
C = plt.contour(X, Y, f(X, Y), 8, colors='black', linewidth=.5)
```

##  Imshow Plots


```python
def f(x, y):
    return (1 - x / 2 + x ** 5 + y ** 3) * np.exp(-x ** 2 - y ** 2)
n = 10
x = np.linspace(-3, 3, 4 * n)
y = np.linspace(-3, 3, 3 * n)
X, Y = np.meshgrid(x, y)
plt.imshow(f(X, Y))
plt.show()
```

```python
fig, ax = plt.subplots()


x_pos = 0
y_pos = 0
x_direct = 1
y_direct = 1


ax.quiver(x_pos, y_pos, x_direct, y_direct)
ax.set_title('Quiver plot with one arrow')


plt.show()
```

###  Quiver plot with two arrows

```python
fig, ax = plt.subplots()

x_pos = [0, 0]
y_pos = [0, 0]
x_direct = [1, 0]
y_direct = [1, -1]


ax.quiver(x_pos,y_pos,x_direct,y_direct,
         scale=5)
ax.axis([-1.5, 1.5, -1.5, 1.5])


plt.show()
```

##  3 D Plots 

```python
from mpl_toolkits.mplot3d import Axes3D
fig = plt.figure()
ax = Axes3D(fig)
X = np.arange(-4, 4, 0.25)
Y = np.arange(-4, 4, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)
ax.plot_surface(X, Y, Z, rstride=1, cstride=1, cmap='hot')
plt.show()
```

## 4.11 Polar Axis

```python
#Polar Axis 
r = np.arange(0, 2, 0.01)
theta = 2 * np.pi * r

ax = plt.subplot(111, projection='polar')
ax.plot(theta, r)
ax.set_rmax(2)
ax.set_rticks([0.5, 1, 1.5, 2])  # less radial ticks
ax.set_rlabel_position(-22.5)  # get radial labels away from plotted line
ax.grid(True)

ax.set_title("A line plot on a polar axis", va='bottom')
plt.show()
```
